```python
# Preprocessing corpus

corpus = "<s> I am Sam. </s>" "<s> Sam I am. </s>" "<s> I do not like green eggs and ham. </s>"

# Tokenize the corpus
tokens = corpus.replace(".", "").replace(",", "").split()
```

```python
# Probability of a word given the previous word

def bigram_probability(word1, word2, tokens):
    bigram_count = 0
    unigram_count = 0

    for i in range(len(tokens) - 1):
        if tokens[i] == word1:
            unigram_count += 1
            if tokens[i + 1] == word2:
                bigram_count += 1

    if unigram_count == 0:
        return 0.0
    return bigram_count / unigram_count

print("P(am | I):", bigram_probability("I", "am", tokens))
print("P(Sam | am):", bigram_probability("am", "Sam", tokens))

```
