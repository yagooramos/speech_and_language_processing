```python
# 1. 
user=input("Write...")
print (len(user))
```

```python
# 2. 
uppercase=user.upper()
print(uppercase)
```

```python
# 3.
def only_alphabets(user):
    return user.isalpha()

if only_alphabets(user):
    print("The string contains only alphabets")
else:
    print("The string does not contain only alphabets")
```

```python
# 4.
reversed_string=""
for char in user:
    reversed_string=char+reversed_string
print(reversed_string)
```

```python
# 5. 
count_repeat=0
for char in user:
    if char=="o":
        count_repeat+=1
print(count_repeat)
```

```python
# 6. 
user1=input("Write again...")
def are_anagrams(user, user1):
    return sorted(user)==sorted(user1)
if are_anagrams(user, user1):
    print("The strings are anagrams")
else:
    print("The strings are not anagrams")
```

```python
# 7.
def remove_vowels(user):
    vowels="aeiouAEIOU"
    no_vowels=""
    for char in user:
        if char not in vowels:
            no_vowels+=char
    return no_vowels
print(remove_vowels(user))
```

```python
# 8.
def palindrome_check(user):
    return user==user[::-1]
if palindrome_check(user):
    print("The strings are palindromes")
else:
    print("The strings are not palindromes")
```

```python
# 9.
first_non_repeated_character=""
char_count={}
for char in user:
    if char in char_count:
        char_count[char]+=1
    else:
        char_count[char]=1
        
    if char_count[char]==1:
        first_non_repeated_character=char
        break
print(first_non_repeated_character)
```

```python
# 10. 
def capitalize_first_letter(user):
    words=user.split()
    capitalized_words=[]
    for word in words:
        capitalized_words.append(word.capitalize())
    return ' '.join(capitalized_words)
print(capitalize_first_letter(user))
```

```python
# 1.
squares = {}
for i in range(1, 11):
    squares[i] = i ** 2
print(squares)
```

```python
# 2.
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
merged_dict = dict1.copy()
merged_dict.update(dict2)
print(merged_dict)
```

```python
# 3.
def count_words(sentence):
    word_count = {}
    words = sentence.split()
    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1
    return word_count

sentence = input("Enter a sentence: ")
print(count_words(sentence))
```

```python
# 4.
keys = ['name', 'age', 'city']
values = ['Yago', 20, 'Santiago']
my_dict = {}
for i in range(len(keys)):
    my_dict[keys[i]] = values[i]
print(my_dict)
```

```python
# 5.
my_dict = {'name': 'Yago', 'age': 20, 'city': 'Santiago'}
key_to_check = input("Enter a key to check: ")
if key_to_check in my_dict:
    print(f"The key '{key_to_check}' exists in the dictionary")
else:
    print(f"The key '{key_to_check}' does not exist in the dictionary")
```

```python
# 6.
def invert_dictionary(my_dict):
    inverted = {}
    for key, value in my_dict.items():
        inverted[value] = key
    return inverted

my_dict = {'a': 1, 'b': 2, 'c': 3}
print(invert_dictionary(my_dict))
```

```python
# 7.
my_dict = {'a': 10, 'b': 25, 'c': 15, 'd': 30}
max_key = None
max_value = None
for key, value in my_dict.items():
    if max_value is None or value > max_value:
        max_value = value
        max_key = key
print(f"Key with highest value: {max_key} = {max_value}")
```

```python
# 8.
def sort_dict_by_values(my_dict):
    sorted_items = sorted(my_dict.items(), key=lambda x: x[1])
    sorted_dict = {}
    for key, value in sorted_items:
        sorted_dict[key] = value
    return sorted_dict

my_dict = {'a': 10, 'b': 5, 'c': 20, 'd': 15}
print(sort_dict_by_values(my_dict))
```

```python
# 9.
my_list = [1, 2, 3, 2, 4, 3, 5, 1]
unique_list = []
for item in my_list:
    if item not in unique_list:
        unique_list.append(item)
print(unique_list)
```

```python
# 10.
my_list = [1, 2, 3, 2, 4, 3, 5, 1]
unique_set = set(my_list)
print(unique_set)
```
