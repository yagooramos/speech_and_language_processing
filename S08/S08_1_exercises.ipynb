### Exercise 3 (Medium)
Build an LSTM-based sentiment classifier.

```python
class LSTMClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim, num_classes):
        super().__init__()
        # Define: embedding, lstm, fc
        embed_dim = 10
        hidden_dim = 20
        lst_dim = 20
        fc_dim = num_classes
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.lstm = nn.LSTM(embed_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(lst_dim, fc_dim)
    
    def forward(self, x):
        # Embed -> LSTM -> take last hidden -> classify
        x = self.embedding(x)
        lstm_out, (h_n, c_n) = self.lstm(x)
        x = self.fc(h_n[-1])
        return x

# Test with dummy data
vocab_size = 100
test_input = torch.randint(0, vocab_size, (1, 5))  # batch=1, seq_len=5
model = LSTMClassifier(vocab_size=100, embed_dim=10, hidden_dim=20, num_classes=2)
output = model(test_input)
print(output)
```
